## 实验1科学计算库Numpy的使用

```python
import numpy as np
```

###### 1.创建一个全零的二维数组，数组的shape为(3,4)。

```python
z1 = np.zeros((3,4))
print (z1)
```

###### 2.创建一个10-49的一维数组，并将其倒序排列。

```python
z2 = np.arange(10,50,1)
z2 = z2[::-1]
z2
```

###### 3.创建二维数组[[1,2,3],[4,5,6]]，并对数组沿着axis=0的轴进行元素求和。

```python
z3 = np.array([[1,2,3],[4,5,6]])
z3
```

```python
np.sum(z3,axis=0)
```

###### 4.创建二维数组[[11,2,3],[4,5,6]]，并对数组沿着axis=1的轴进行元素累乘。

```python
z4 = np.array([[11,2,3],[4,5,6]])
z4
```

```python
z4.prod(axis = 1)
```

###### 5.随机创建一个二维数组，数组的shape为(3,4)，并打印数组、最大值和最大值的索引。

```python
z5 = np.random.random((3,4))
print(z5)
print(z5.min())
print(z5.argmin())
```

###### 6.随机创建一个二维数组，数组的shape为(3,4)，并对数组做归一化操作。

```python
z6 = np.random.random((3,4))
z6_max = z6.max()
z6_min = z6.min()
z6 = (z6-z6_min)/(z6_max - z6_min)
z6
```

###### 7.创建二维数组[[1.5,1.3,7.5],[5.6,7.8,1.2]]，并对数组每列上的元素进行升序排序。

```python
z7 = np.array([[1.5,1.3,7.5],
               [5.6,7.8,1.2]])
```

```python
np.sort(z7,axis = 0)
```
