# 实验3：数据可视化库Matplotlib的使用

### 1.导入matplotlib库简写为plt

```python
import matplotlib.pyplot as plt
import numpy as np
```

### 2.用scatter方法画出x=(0,10)间sin的点图像

```python
x = np.linspace(0, 10, 30)
plt.scatter(x, np.sin(x));
```

### 3.绘制一组误差为±0.8的数据的误差条图

```python
x = np.linspace(0, 10, 50)
dy = 0.8
y = np.sin(x) + dy * np.random.randn(50)

plt.errorbar(x, y, yerr=dy, fmt='.k')
```

```python
x = [1,2,3,4,5,6,7,8]
y = [3,1,4,5,8,9,7,2]
label=['A','B','C','D','E','F','G','H']

plt.bar(x,y,tick_label = label);
```

### 5.绘制一个水平方向柱状图

### 6.绘制1000个随机值的直方图

```python
data = np.random.randn(1000)
plt.hist(data);
```

### 7.设置直方图分30个bins，并设置为频率分布

```python
plt.hist(data, bins=30,histtype='stepfilled', density=True)
plt.show();
```

### 8.在一张图中绘制3组不同的直方图，并设置透明度

```python
x1 = np.random.normal(0, 0.8, 1000)
x2 = np.random.normal(-2, 1, 1000)
x3 = np.random.normal(3, 2, 1000)

kwargs = dict(alpha=0.3, bins=40, density = True)

plt.hist(x1, **kwargs);
plt.hist(x2, **kwargs);
plt.hist(x3, **kwargs);
```

### 9.绘制x=(0,10)间sin的图像；
### 设置线性为虚线；
### 设置y轴显示范围为(-1.5,1.5)；
### 设置x,y轴标签variable x，value y；
### 设置图表标题“sin”；
### 显示网格。
### 绘制平行于x轴y=0.8的水平参考线。

```python
x = np.linspace(0,10,100)
plt.plot(x,np.sin(x),'--');
plt.ylim(-1.5, 1.5);
plt.xlabel('variable x');
plt.ylabel('value y');
plt.title('sin');
plt.grid()
plt.axhline(y=0.8, ls='--', c='r')
```

### 10.绘制x=(0,10)间sin的图像，并绘制垂直于x轴x<4 and x>6的参考区域，以及y轴y<0.2 and y>-0.2的参考区域

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.axvspan(xmin=4, xmax=6, facecolor='r', alpha=0.3) # 垂直x轴
plt.axhspan(ymin=-0.2, ymax=0.2, facecolor='y', alpha=0.3);  # 垂直y轴
```

### 11.绘制x=(0,10)间sin的图像，并添加注释文字sin(x)

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.text(3.2, 0, 'sin(x)', weight='bold', color='r');
```

### 12.绘制x=(0,10)间sin的图像，并绘制箭头标出第一个峰值

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.annotate('maximum',xy=(np.pi/2, 1),xytext=(np.pi/2+1, 1),
             weight='bold',
             color='r',
             arrowprops=dict(arrowstyle='->', connectionstyle='arc3', color='r'));
```

### 13.在一张图里绘制sin,cos的图形，并展示图例；
###     调整图例在右上角展示，且不显示边框；

```python
x = np.linspace(0, 10, 1000)
fig, ax = plt.subplots()

ax.plot(x, np.sin(x), label='sin')
ax.plot(x, np.cos(x), '--', label='cos')
ax.legend(loc='upper right', frameon=False);
```

### 14.在2个子图中，显示sin(x)和cos(x)的图像

```python
fig = plt.figure()
ax1 = fig.add_axes([0.1, 0.5, 0.8, 0.4], ylim=(-1.2, 1.2))
ax2 = fig.add_axes([0.1, 0.1, 0.8, 0.4], ylim=(-1.2, 1.2))

x = np.linspace(0, 10)
ax1.plot(np.sin(x));
ax2.plot(np.cos(x));
```

### 15.显示一组二维数据的频度分布，并分别在x,y轴上，显示该维度的数据的频度分布

```python
mean = [0, 0]
cov = [[1, 1], [1, 2]]
x, y = np.random.multivariate_normal(mean, cov, 3000).T

# Set up the axes with gridspec
fig = plt.figure(figsize=(6, 6))
grid = plt.GridSpec(4, 4, hspace=0.2, wspace=0.2)
main_ax = fig.add_subplot(grid[:-1, 1:])
y_hist = fig.add_subplot(grid[:-1, 0], xticklabels=[], sharey=main_ax)
x_hist = fig.add_subplot(grid[-1, 1:], yticklabels=[], sharex=main_ax)

# scatter points on the main axes
main_ax.scatter(x, y,s=3,alpha=0.2)

# histogram on the attached axes
x_hist.hist(x, 40, histtype='stepfilled',
            orientation='vertical')
x_hist.invert_yaxis()

y_hist.hist(y, 40, histtype='stepfilled',
            orientation='horizontal')
y_hist.invert_xaxis()
```

### 16.绘制一组三维点

```python
ax = plt.axes(projection='3d')
zdata = 15 * np.random.random(100)
xdata = np.sin(zdata) + 0.1 * np.random.randn(100)
ydata = np.cos(zdata) + 0.1 * np.random.randn(100)
ax.scatter3D(xdata, ydata, zdata, c=zdata, cmap='Greens');
```

## 17、宝可梦数据集可视化

```python
import pandas as pd
df = pd.read_csv('Pokemon.csv')
```

### (1).用scatter的x,y,c属性，展示所有宝可梦的Defense,Attack,HP数据

```python
x = df['Attack']
y = df['Defense']
colors = df['HP']

plt.scatter(x, y, c=colors, alpha=0.5)
plt.title('Scatter plot')
plt.xlabel('HP')
plt.ylabel('Attack')
plt.colorbar();
```

### (2).展示所有宝可梦的攻击力的分布直方图，bins=10

```python
x = df['Attack']
num_bins = 10
n, bins, patches = plt.hist(x, num_bins, facecolor='blue', alpha=0.5)
plt.title('Histogram')
plt.xlabel('Attack')
plt.ylabel('Value')
plt.show()
```
