### 5. 定义df：从字典对象创建DataFrame，并设置索引

```python
import numpy as np
data = {
    'animal':
    ['cat', 'cat', 'snake', 'dog', 'dog', 'cat', 'snake', 'cat', 'dog', 'dog'],
    'age': [2.5, 3, 0.5, np.nan, 5, 2, 4.5, np.nan, 7, 3],
    'visits': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'priority':
    ['yes', 'yes', 'no', 'yes', 'no', 'no', 'no', 'yes', 'no', 'no']
}

labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
df = pd.DataFrame(data, index=labels)
df
```

### 5.1 展示df的前3行

```python
df.iloc[:3]
# 方法二
#df.head(3)
```

### 5.2 取出df的animal和age列

```python
df.loc[:, ['animal', 'age']]
# 方法二
# df[['animal', 'age']]
```

### 5.3 取出索引为[3, 4, 8]行的animal和age列

```python
df.loc[df.index[[3, 4, 8]], ['animal', 'age']]
```

### 5.4. 取出age值大于3的行

```python
df[df['age'] > 3]
```

### 5.5 取出age值缺失的行

```python
df[df['age'].isnull()]
```

### 5.6 取出age在2,4间的行（不含）

```python
df[(df['age']>2) & (df['age']>4)]
# 方法二
# df[df['age'].between(2, 4)]
```

### 5.7 f行的age改为1.5

```python
df.loc['f', 'age'] = 1.5
```

### 5.8 计算visits的总和

```python
df['visits'].sum()
```

### 5.9 计算每个不同种类animal的age的平均数

```python
df.groupby('animal')['age'].mean()
```

### 5.10 在df中插入新行k，df.loc['k'] = [5.5, 'dog', 'no', 2]，然后删除该行

```python
#插入
df.loc['k'] = [5.5, 'dog', 'no', 2]
# 删除
df = df.drop('k')
df
```

### 5.11 计算df中每个种类animal的数量

```python
df['animal'].value_counts()
```

### 5.12 先按age降序排列，后按visits升序排列

```python
df.sort_values(by=['age', 'visits'], ascending=[False, True])
```

### 5.13. 将priority列中的yes, no替换为布尔值True, False

```python
df['priority'] = df['priority'].map({'yes': True, 'no': False})
df
```

### 5.14 对每种animal的每种不同数量visits，计算平均age，即，返回一个表格，行是aniaml种类，列是visits数量，表格值是行动物种类列访客数量的平均年龄

```python
df.pivot_table(index='animal', columns='visits', values='age', aggfunc='mean')
```
